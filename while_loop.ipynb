## python loop
#### while loop **condition statement**
## while\<condition>:


```python
a=1
while a<=10:
    print(a,end="")
    a+=1
print("\nloop over")
```

```python
a = 0
while a<=100:
    print(a,end="")
    a+=10
print("\nloop over")
```

```python
a = 100
while a>=0:
    print(a,end="")
    a-=10
print("\nloop over")
```

#### break-stop the loop

```python
a= 1
while a>=0:
    if a == 20:
        break
    print(a,end=",")
    a+=1
print("\nloop ends")
```

#### continue - go back to the top of my loop

```python
a= 1
while a<=20:
    if a%3 ==0:
        
        a+=1
        continue
    print(a,end=",")
    a+=1
```

#### nested while loop

```python
a =1
while a<=5:
    b =1
    while b<=5:

        print(b,end="")

        b+=1

    print()
    a+=1
print("\nloop over")
```

```python
row=int(input("enter the row"))
column=int(input("enter the column"))
a=1
while a<=row:
    b=1
    while b<=column:
        print(b,end="")
        b+=1
    print()
    a+=1
print("loop over")

```

#### for loop 
**range()**     
syntax  
**for \<variable_name> in \<str\list\tuple\set\dict>:**

```python
for i in ("hello world"):
    print(i,end="")
print("\nloop over")
```

```python
mylist = [100,"hello","vicky","hi",25.00,True]
for i in mylist:
    print(i)
```

```python
# range
for i in range(20):
    print(i,end=",")
```

```python
for i in range(50,100+1):
    print(i,end=",")
```

```python
for i in range(0,100+1,10):
    print(i,end=",")
```

```python
for i in range (100,0,-10):
    print(i,end=",")
```

```python
# for with else

for i in "hello vicky":
    print(i,end="")
else:
    print("\nsuccess")
```
