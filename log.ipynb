```python
import os
import csv
profile_name =[]
movie_name =[]



path ="C:\\Users\\VIGNESH\\OneDrive\\Documents\\csv"
filename ="All_SearchHistory.csv"
print(os.path.join(path,filename))
with open (os.path.join(path,filename)) as f:
    for each in f:
        mylist = each.split(",")
        profile_name.append(mylist[0])
        movie_name.append(mylist[5])
print(profile_name)
print(movie_name)

def frequency(mylist):
    mydict={}
    for each in mylist:
        if each not in mydict:
            mydict[each] =1
        else:
            mydict[each] +=1
    return mydict

def to_csv(mydict):
    with open(os.path.join(path,"new.csv"),"w",newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["sno","profile_name", "frequency"])
        sno=1
        for each in mydict:
            writer.writerow([sno,each,mydict[each]])
            sno+=1
x=frequency(profile_name)
print(x)

to_csv(x)



           


```
