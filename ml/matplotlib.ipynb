#### **scatter plot**

```python
import numpy as np
x = np.array([1,2,3])
y = np.array([1,2,3])
plt.scatter(x,y,c=["r","g","b"],marker="*",s=200)
plt.show();
```

### **line plot**

```python
x = np.array([1,2,3,4,5])
y = np.array([1,4,3,5,6])
C=["red","cyan","yellow","blue","olive"]
plt.plot(x,y,colors=C,marker="*",linestyle="dashed")
plt.show();
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:\\Users\\VIGNESH\\OneDrive\\Documents\\genai-python\\ml\\heart-disease.csv")

plt.scatter(df["sex"], df["chol"])

mean_target = df["target"].mean()
plt.axhline(y=mean_target, color="red", linestyle="--",
            label=f"Mean Target = {mean_target:.2f}")

plt.xlabel("Sex")
plt.ylabel("chol")
plt.title("Sex vs Target")
plt.legend()
plt.show()
```
