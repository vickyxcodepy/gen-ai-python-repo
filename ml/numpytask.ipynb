```python
df = pd.DataFrame(data)
df
```

```python
#1
print(df.head())
print(df.tail(3))
print(df.shape)
print(df.dtypes)
```

```python
#2
result = df[(df["Department"] == "IT") & (df["Salary"] > 40000)]
result
```

```python
#3
result = df.sort_values(by = "Salary",ascending=False)
print(result[["Name","Department","Salary"]])
```

```python
#4
print("means Salary : ",df["Salary"].mean())
print("max Salary : ",df["Salary"].max())
print("min Salary : ",df["Salary"].min())
print("average age : ",df["Age"].mean())
```

```python
#5
result = df.groupby("Department").agg(
    EmployeeCount=("EmpID", "count"),
    AverageSalary=("Salary", "mean"),
    HighestSalary=("Salary", "max")
)

result = result.sort_values(
    by="AverageSalary",
    ascending=False
)

print(result)
```

```python
#6
df["Annualsalary"] = df["Salary"] *12
print(df[["Name","Salary","Annualsalary"]])
```

```python
#7
df_copy = df.copy()


df_copy.loc[0, "City"] = None


df_copy = pd.concat(
    [df_copy, df_copy.iloc[[1]]],
    ignore_index=True
)

print("Missing values:")
print(df_copy.isnull().sum())


df_copy["City"] = df_copy["City"].fillna("Unknown")


df_copy = df_copy.drop_duplicates().reset_index(drop=True)

print("\nCleaned DataFrame:")
print(df_copy)

print("\nRemaining duplicates:", df_copy.duplicated().sum())
```

```python
#8
conditions = [
    df["Salary"] >= 50000,
    (df["Salary"] >= 40000) & (df["Salary"] < 50000),
    df["Salary"] < 40000
]

choices = ["High", "Medium", "Low"]

df["SalaryLevel"] = pd.Series(
    pd.NA, index=df.index, dtype="string"
)

import numpy as np

df["SalaryLevel"] = np.select(
    conditions,
    choices,
    default="Low"
)

print(df[["Name", "Salary", "SalaryLevel"]])
```

```python
#9
marks = pd.DataFrame({
    "EmpID": [101, 103, 105, 108, 110],
    "TrainingScore": [85, 92, 78, 88, 90]
})

result = pd.merge(
    df,
    marks,
    on="EmpID",
    how="left"
)

print(result[["EmpID", "Name", "TrainingScore"]])
```

```python
#10
department_avg = df.groupby("Department")["Salary"].mean()

highest_department = department_avg.idxmax()
highest_average = department_avg.max()

print("Department with highest average salary:", highest_department)
print("Highest department average salary:", highest_average)


overall_average = df["Salary"].mean()

above_average = df[df["Salary"] > overall_average]

print("\nOverall average salary:", overall_average)
print("\nEmployees earning above average:")
print(above_average[["Name", "Department", "Salary"]])
```
